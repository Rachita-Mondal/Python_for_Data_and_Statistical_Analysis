## Lecture Section

In this lecture, we will cover some of the `NumPy` library.
Specifically:
* Data manipulation
    * Arrays
    * Shape & Dimension
    * Types & Conversions
    * Sorting\Filtering\Searching

NumPy is short for "Numerical Python". It is a library specifically made for working with arrays which are list-like objects.

Arrays are quicker to access and manipulate than lists because they are stores in one continuous place in memory. Since NumPy is an open-source project, it is frequently updated and optimized for new systems and architecture, too.

Similar to Pandas, we use an alias `np` for NumPy.

```python
import numpy as np
```

### Data Manipulation

To create an array, we call `np.array()` and pass a list or tuple as argument.

```python
arr = np.array([1, 2, 3, 4, 5])
arr
```

```python
type(arr)
```

Each item in an array is a 0-D array, or a **Scalar**. Above, we create a 1-D scalar. We can also create nested arrays that have higher dimensions. These arrays have other 1-d, 2-D, etc. arrays as their items.

```python
arr_2d = np.array([[1, 2, 3], [4, 5, 6]]) # commonly a matrix
arr_2d
```

```python
arr_3d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
arr_3d
```

We can check the number of dimensions an array has with `.ndim`.

```python
arr.ndim
```

```python
arr_2d.ndim
```

This related to the **shape** of the array, which is the number of elements in each dimension.

```python
arr.shape
```

```python
arr_3d.shape
```

We can reshape, too

```python
arr_reshape = arr_3d.reshape(12)
arr_reshape
```

```python
arr_reshape.reshape(4, 3)
```

We index an array the exact same we would a list. Make sure you understand how the indexing working below.

```python
arr_3d[1, 0, 2]
```

Similar to lists and strings, we can slice the arrays. As a refresher, we can slice in two ways:

`[start:end]`
`[start:end:step]`

* *start* is the index we begin at and is included - defaults to 0
* *end* is the index we end at and is not included - defaults to the length of the array
* *step* how many items we iterate by - defaults to 1

```python
arr[1:2]
```

```python
arr_3d[1, 0, 1:3]
```

We can check the type of the element sin the array using `.dtype`, and we can convert using `.astype()`

```python
arr.dtype
```

```python
print(arr.astype('bool'))
```

We can make a copy of the array we have. This creates a new object, so changing the copy will not change the original.

```python
arr2 = arr.copy()
arr2[0] = 100
print(arr)
print(arr2)
```

Iterating through an array is also the same as it is for lists. However, there is an easier method of getting the **Scalar** values from a nested array, rather than using nested for-loops. You can use `np.nditer()`. For enumerating, you may use `np.ndenumerate()`.

Both methods are shown below.

```python
for x in arr_3d: # a for-loop for each dimension
    for y in x:
        for z in y:
            print(z)
```

```python
for x in np.nditer(arr_3d): # creates the for-loops for you!
    print(x)
```

There are several ways to join NumPy arrays.
* `np.concatenate()`
* `np.stack()`

These functions take an **axis** argument. View this webpage for a quick explanation on axes.

https://www.sharpsightlabs.com/blog/numpy-axes-explained/

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

np.concatenate((arr1, arr2)) # axis defaults to 0
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

np.concatenate((arr1, arr2), axis=1)
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

np.stack((arr1, arr2), axis=1)
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

np.hstack((arr1, arr2))  # along rows
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

np.vstack((arr1, arr2)) # along columns
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

np.dstack((arr1, arr2)) # along height
```

We can search in an array using `np.where()`

```python
np.where(arr == 4)
```

Sorting happens with `np.sort()`

```python
arr3 = np.array([3, 1, 2])
np.sort(arr3)
```

To flip the order, we can do a slice with the **step** set to **-1** (this flips whatever we are slicing - strings, too!)

```python
np.sort(arr3)[::-1]
```

Finally, we can create a filter array of boolean values to filter our arrays. Each index that has `True` is kept in the original array.

```python
arr4 = np.array([1, 2, 3, 4, 5])
filter_list = [True, False, True, True, False]

arr4[filter_list]
```

```python
filter_list = []
for element in arr4:
    if element % 2 == 0: # modulus operator, returns remainder after integer division
        filter_list.append(True)
    else:
        filter_list.append(False)

arr4[filter_list]
```

**Question 1.** For this question, you will create a function called `stacking()`
* Parameters: 4. THe first two are arrays, the third (option) and fourth (axis) are integers.
* Returns: A stacked array.
* Goal: The function will take two arrays and stack them based on the option selected, and the axis if needed. Axis should default to 0 and is only needed for the first two options.
    * Option 1: concatenate (requires axis)
    * Option 2: stack (requires axis)
    * Option 3: hstack
    * Option 4: vstack
    * Option 5: dstack
* Example: `stacking(arr1, arr2, 1)` will concatenate `arr1` and `arr2` with an axis of `0`

```python
def stacking(arr1, arr2, option, axis = 0):
    if option == 1:
        result = np.concatenate((arr1, arr2), axis)
    elif option == 2:
        result = np.stack((arr1, arr2), axis)
    elif option == 3:
        result = np.hstack((arr1, arr2))
    elif option == 4:
        result = np.vstack((arr1, arr2))
    elif option == 5:
        result = np.dstack((arr1, arr2))
    return result
```
